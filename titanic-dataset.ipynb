INDEXING, SELECTING, FILTRING

Select only the columns Name, Sex and Age, using loc[]

```python
df.loc[:, ['Name', 'Sex', 'Age']]
```

Select the first 10 rows and the first 4 columns using iloc[]

```python
df.iloc[:10, :4]
```

Filter df to show only female passengers. Use loc[] to select where sex == female

```python
df.loc[df.Sex == 'female']
```

Filter df to show passenger from Pclass == 1 e older than 40

```python
df.loc[(df.Pclass == 1) & (df.Age > 40)]
```
